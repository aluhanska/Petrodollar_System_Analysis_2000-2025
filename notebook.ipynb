## 📌 Executive Summary

Нафтодоларова система — це механізм, через який нафтові доходи країн-експортерів повертаються у долар США як домінантну валюту глобальних розрахунків. Цей проєкт перевіряє, наскільки цей механізм ще працює, на основі **щоденних цін, виробничих даних, рішень [ОПЕК](https://uk.wikipedia.org/wiki/%D0%9E%D0%9F%D0%95%D0%9A) і метрик [дедоларизації](https://uk.wikipedia.org/wiki/%D0%94%D0%B5%D0%B4%D0%BE%D0%BB%D0%B0%D1%80%D0%B8%D0%B7%D0%B0%D1%86%D1%96%D1%8F) за 2000–2025 роки**.

**Три головних висновки наперед:**
1. ⚠️ Кореляція між дотриманням квот ОПЕК і ціною нафти **статистично незначуща** (r ≈ 0.06, p > 0.05) — картель уже не контролює ринок як раніше.
2. 💵 Долар і нафта пов'язані обернено (r = -0.45, p < 0.001) — [DXY](https://uk.wikipedia.org/wiki/%D0%86%D0%BD%D0%B4%D0%B5%D0%BA%D1%81_%D0%B4%D0%BE%D0%BB%D0%B0%D1%80%D0%B0) залишається важливим макроіндикатором.
3. 🐉 Дедоларизація **реальна, але повільна**: частка USD у глобальних резервах впала з 66.5 % до 56.8 % за 25 років, а [CAGR](https://uk.wikipedia.org/wiki/CAGR) нафтоюаня — 59 % на рік після 2018.

### Питання 1 — Чи дійсно ОПЕК контролює ціну нафти через дотримання квот?

**Бізнес-логіка:** Якщо ОПЕК має реальну ринкову силу, періоди високого дотримання квот мають збігатися з вищими цінами Brent.

- **H₀:** кореляція дотримання квот ОПЕК ↔ середня ціна Brent = 0 (відсутній зв'язок)
- **H₁:** кореляція статистично значуща (α = 0.05)


```python
# Агрегація: дотримання квот ОПЕК та ціни Brent по роках
sql_q1 = f"""
SELECT
    year,
    ROUND(AVG(opec_compliance_pct), 2) AS avg_compliance_pct,
    ROUND(AVG(brent_price_usd), 2) AS avg_brent_usd,
    ROUND(AVG(opec_actual_production_mbd), 2) AS avg_opec_production_mbd
FROM {full_table_name('opec_quotas_events')}
GROUP BY year
ORDER BY year
"""

q1_yearly = run_query(sql_q1)
q1_yearly.head()
```

```python
# Статистичний тест: кореляція Пірсона
r_q1, p_q1 = stats.pearsonr(
    df_opec["opec_compliance_pct"], df_opec["brent_price_usd"]
)

print(f"Кореляція Pearson: r = {r_q1:.3f}")
print(f"p-value:           {p_q1:.4f}")
print()

if p_q1 < 0.05:
    print("✅ Відхиляємо H₀: кореляція статистично значуща")
else:
    print("❌ Не відхиляємо H₀: значущого зв'язку не виявлено")
```

```python
# Line chart: дотримання квот ОПЕК vs середня ціна Brent по роках
# з подвійною віссю
fig, ax1 = plt.subplots(figsize=(14, 5))

color_brent = "#1565C0"
color_compl = "#FF6F00"

ax1.fill_between(q1_yearly["year"], q1_yearly["avg_brent_usd"],
                 alpha=0.15, color=color_brent)
ax1.plot(q1_yearly["year"], q1_yearly["avg_brent_usd"],
         color=color_brent, linewidth=2.5, label="Brent (USD/bbl)")
ax1.set_xlabel("Рік")
ax1.set_ylabel("Середня ціна Brent, USD/bbl", color=color_brent)
ax1.tick_params(axis="y", labelcolor=color_brent)

ax2 = ax1.twinx()
ax2.plot(q1_yearly["year"], q1_yearly["avg_compliance_pct"],
         color=color_compl, linewidth=2, linestyle="--",
         marker="o", markersize=4, label="Дотримання квот ОПЕК, %")
ax2.axhline(100, color="gray", linestyle=":", linewidth=1, alpha=0.6)
ax2.set_ylabel("Дотримання квот ОПЕК, %", color=color_compl)
ax2.tick_params(axis="y", labelcolor=color_compl)

ax1.set_title(
    f"Дотримання квот ОПЕК vs середня ціна Brent "
    f"(r = {r_q1:.2f}, p = {p_q1:.2f})"
)
ax1.grid(False)
ax2.grid(False)
fig.tight_layout()
plt.show()
```

**Спостереження:** Кореляція виявилась **дуже слабкою (r = 0.06) і статистично незначущою (p = 0.27)**. Це переломне відкриття: попри загальне сприйняття ОПЕК як головного регулятора нафтового ринку, **сам факт виконання квот картелем не пояснює рух цін**. На графіку видно, що дані щодо дотримання квот коливаються у вузькому діапазоні 97.5–101 % — члени ОПЕК загалом дотримуються домовленостей. Але ціна Brent живе своїм життям: у 2008 році дотримання квот впало до мінімуму ~97.3 %, а ціна одночасно злетіла до \$97/bbl; у 2014 при нормальному дотриманні ~98.3 % ціна обвалилась з \$98 до \$52 через сланцеву революцію США; у 2020 ціна впала через COVID-19 попри стабільне дотримання ~98 %; у 2022 і дотримання, і ціна зросли разом — але причиною був не картель, а вторгнення в Україну. Ринок реагує на ширшу систему чинників — попит Китаю, сланцеву пропозицію США, геополітичні шоки — а не лише на дисципліну ОПЕК.

### Питання 2 — Як сланцева революція США змінила глобальний баланс виробництва?

**Бізнес-логіка:** Поява нового великого виробника поза картелем має зменшити концентрацію виробництва і ослабити вплив ОПЕК.


```python
# Розрахунок зростання сланцю: спочатку рахуємо річне сланцеве виробництво,
# потім додаємо лагові показники з попереднього року
sql_q2 = f"""
WITH yearly_shale AS (
    SELECT
        year,
        ROUND(AVG(us_shale_production_mbd), 2) AS shale_mbd,
        ROUND(AVG(non_opec_plus_production_mbd), 2) AS non_opec_plus_mbd,
        ROUND(AVG(opec_actual_production_mbd), 2) AS opec_mbd
    FROM {full_table_name('opec_quotas_events')}
    GROUP BY year
)
SELECT
    year,
    shale_mbd,
    opec_mbd,
    non_opec_plus_mbd,
    ROUND(
        shale_mbd - LAG(shale_mbd) OVER (ORDER BY year),
        2
    ) AS shale_growth_yoy
FROM yearly_shale
ORDER BY year
"""

q2_shale = run_query(sql_q2)
q2_shale.head(10)
```

```python
# Зростання сланцевого виробництва: точка старту, пік
shale_2007 = q2_shale.loc[q2_shale["year"] == 2007, "shale_mbd"].iloc[0]
shale_peak = q2_shale["shale_mbd"].max()
shale_peak_year = int(q2_shale.loc[q2_shale["shale_mbd"].idxmax(), "year"])

print(f"Сланцеве виробництво США, 2007:   {shale_2007:.2f} Mb/d")
print(f"Пік сланцевого виробництва, {shale_peak_year}: {shale_peak:.2f} Mb/d")
print(f"Сукупний приріст:                 +{shale_peak - shale_2007:.2f} Mb/d")
```

Стратегія провалилася: американський сланець виявився **стійкішим, ніж очікувалося** завдяки технологічним вдосконаленням і зниженню собівартості.


### Питання 3 — Які роки спричинили найбільшу волатильність цін Brent?

**Бізнес-логіка:** Коефіцієнт варіації (CV = std / mean × 100 %) — індикатор нестабільності ціни. Високий CV сигналізує про шок у відповідному році.


```python
# Волатильність Brent по роках через коефіцієнт варіації
price_volatility = (
    df_prices.groupby("year")["brent_crude_usd"]
    .agg(mean_price="mean", std_price="std")
    .assign(cv_pct=lambda x: x["std_price"] / x["mean_price"] * 100)
    .round(2)
    .sort_values("cv_pct", ascending=False)
)

print("Топ-5 років за коефіцієнтом варіації Brent:")
price_volatility.head()
```

```python
# Економічні кризи нафтового ринку
CRISIS_YEARS = (2008, 2009, 2015, 2016, 2020)

# Bar chart: коефіцієнт варіації цін Brent по роках
fig, ax = plt.subplots(figsize=(14, 6))
colors = [
    "#E53935" if y in CRISIS_YEARS else "#1565C0"
    for y in price_volatility.index
]
bars = ax.barh(price_volatility.index.astype(int), price_volatility["cv_pct"],
               color=colors, alpha=0.85)
ax.bar_label(bars, fmt="%.2f", padding=3, fontsize=9)

ax.set_title("Волатильність цін Brent")
ax.set_xlabel("Коефіцієнт варіації, %")
ax.set_ylabel("Рік")

legend_elements = [
    Patch(facecolor="#E53935", alpha=0.85, label="Кризові роки"),
    Patch(facecolor="#1565C0", alpha=0.85, label="Інші"),
]
ax.legend(handles=legend_elements, loc="upper right")
ax.grid(False)
plt.tight_layout()
plt.show()
```

**Спостереження:** Найвищі коефіцієнти варіації у 2000-2003 (~7-9 %) при низьких абсолютних цінах. Цікаво, що **навіть «класичні» кризові роки (червоні бари)** не показують найвищої волатильності: 2008 на 3-му місці, 2020 у топ-7, а 2015-2016 — далеко не лідери. Це підтверджує, що **коефіцієнт варіації перебільшує нестабільність дешевих ринків** — у роки \$30/bbl навіть \$3 коливання дає коефіцієнт ~10 %, тоді як ті самі \$3 при \$100/bbl — лише 3 %.

### Питання 4 — Чи існує статистично значущий обернений зв'язок між DXY та Brent?

**Бізнес-логіка:** Оскільки нафта котирується в доларах, сильний долар робить нафту дорожчою для нерезидентних покупців → знижує попит → тисне ціну вниз.

- **H₀:** кореляція DXY ↔ Brent = 0 (відсутній зв'язок)
- **H₁:** кореляція статистично значуща (α = 0.05)


```python
# Статистичний тест: кореляція Пірсона
r_q4, p_q4 = stats.pearsonr(
    df_prices["brent_crude_usd"], df_prices["usd_dxy_index"]
)

print(f"Кореляція Pearson: r = {r_q4:.3f}")
print(f"p-value:           {p_q4:.2e}")
print()

if p_q1 < 0.05:
    print("✅ Відхиляємо H₀: кореляція статистично значуща")
else:
    print("❌ Не відхиляємо H₀: значущого зв'язку не виявлено")
```

```python
# Scatter plot з кольоровою шкалою по роках і лінією регресії
fig, ax = plt.subplots(figsize=(11, 5))

scatter = ax.scatter(
    df_prices["usd_dxy_index"],
    df_prices["brent_crude_usd"],
    c=df_prices["year"],
    cmap="RdYlBu_r",
    alpha=0.3,
    s=10
)

# Лінія лінійної регресії
dxy = df_prices["usd_dxy_index"]
brent = df_prices["brent_crude_usd"]

slope, intercept, r_value, *_ = stats.linregress(dxy, brent)
x_line = np.linspace(dxy.min(), dxy.max(), 200)
ax.plot(x_line, slope * x_line + intercept,
        color="red", linewidth=2.2,
        label=f"Лінія тренду (r = {r_value:.2f})")

plt.colorbar(scatter, ax=ax, label="Рік")
ax.set_title("DXY vs Brent: обернений зв'язок підтверджено")
ax.set_xlabel("Сила долара")
ax.set_ylabel("Ціна Brent, USD/bbl")
ax.legend(loc="upper right")
ax.grid(False)
plt.tight_layout()
plt.show()
```

**Спостереження:** Кореляція **r = -0.45 (p < 0.001)** — це **помірний і високо статистично значущий обернений зв'язок**. Слабкий долар систематично відповідає дорожчій нафті. Але r = -0.45, а не -1 — значить інші фактори (рішення ОПЕК, геополітика, попит Китаю) пояснюють більшу частину варіації ціни. На графіку добре видно три характерні кластери: точки **2000-х років (темно-сині)** зосереджені у нижньому правому куті — високий DXY (110-120) поєднується з дуже низькою ціною (\$20-30/bbl); точки **2008-2014 (блакитно-жовті)** — у верхньому лівому: низький DXY (75-90) і висока ціна (\$90-115/bbl), що класично підтверджує теорію; але **2022-2025 (червоні)** ламають патерн — і DXY високий (100-110), і ціна висока (\$75-105/bbl) одночасно, бо геополітичний шок від війни в Україні переважив валютний канал. **DXY — корисний, але не єдиний індикатор**: коли геополітика домінує, традиційні залежності перестають працювати.


### Питання 5 — Які країни найбільше виграли від нафтового суперциклу 2021–2022?

**Бізнес-логіка:** Суперцикл після COVID-19 і вторгнення росії в Україну вивів Brent до $120+. Порівняння доходів 2019 vs 2022 показує реальних бенефіціарів.


```python
# Агрегація: зведена таблиця
sql_q5 = f"""
SELECT
    country,
    region,
    is_opec_member,
    ROUND(
        SUM(CASE WHEN year = 2019 THEN oil_revenue_usd_bn END), 2
    ) AS revenue_2019,
    ROUND(
        SUM(CASE WHEN year = 2022 THEN oil_revenue_usd_bn END), 2
    ) AS revenue_2022,
    ROUND(
        SUM(CASE WHEN year = 2022 THEN oil_revenue_usd_bn END)
      - SUM(CASE WHEN year = 2019 THEN oil_revenue_usd_bn END), 2
    ) AS growth_usd_bn
FROM {full_table_name('oil_production_trade')}
WHERE year IN (2019, 2022)
GROUP BY country, region, is_opec_member
ORDER BY growth_usd_bn DESC
"""

q5_winners = run_query(sql_q5)
q5_winners.head()
```

### Питання 6 — Як суверенні фонди залежать від ціни нафти?

**Бізнес-логіка:** Суверенні фонди формуються за рахунок нафтових надприбутків — отже їх розмір має корелювати з ціною Brent.

```python
# Кореляційний аналіз
def correlate_with_brent(df: pd.DataFrame, brent_col: str,
                         target_cols: dict[str, str]) -> pd.DataFrame:
    """Обчислити кореляції Pearson між Brent і колонками SWF.

    target_cols: словник {колонка_у_df: відображувана_назва_фонду}
    """
    results = []
    for col, name in target_cols.items():
        r, p = stats.pearsonr(df[brent_col], df[col])
        results.append({
            "Фонд": name,
            "Кореляція Pearson (r)": round(r, 3),
            "p-value": round(p, 4),
        })
    return pd.DataFrame(results)


swf_corr = correlate_with_brent(
    df_swf,
    brent_col="brent_avg_usd",
    target_cols={
        "norway_gpfg_aum_usd_bn": "Government Pension Fund of Norway",
        "saudi_pif_aum_usd_bn": "PIF",
        "adia_aum_usd_bn": "ADIA",
        "kuwait_kia_aum_usd_bn": "KIA",
        "qatar_qia_aum_usd_bn": "QIA",
        "russia_nwf_aum_usd_bn": "Russia NWF",
    },
)
swf_corr
```

Це показує, що **поточна ціна нафти — не єдиний драйвер AUM**: фонди ростуть і завдяки інвестиційному доходу, новим внескам уряду, а також **лагованим ефектам** — нафтові доходи минулих років капіталізуються у фондах роками. На графіку це чітко видно: під час обвалу цін 2014-2016 (з \$110 до \$44) фонди продовжували зростати майже без сповільнення. **Government Pension Fund of Norway** виріс у 10 разів за 25 років (з \$174 до \$1750 млрд) і обігнав ADIA у 2015 році, ставши найбільшим у світі. **ADIA з найвищим значенням r (0.59)** — тому що ОАЕ найдовше акумулюють фонд (стартував з \$270 млрд ще у 2000), а монобізнес-економіка робить фонд найчутливішим до нафтових циклів. **Винятком є Russia NWF** — єдиний фонд, що скорочується: з піку \$185 млрд у 2021 впав до \$120 млрд у 2025 через санкції і покриття бюджетного дефіциту після вторгнення в Україну. Це показує, що суверенні фонди — **не вічні скарбниці**, а інструмент, який можна швидко вичерпати під політичним тиском.


### Питання 7 — Наскільки швидко зростає нафтоюань після запуску INE-ф'ючерсів?

**Бізнес-логіка:** Шанхайська біржа INE запустила нафтові ф'ючерси в юанях у березні 2018 року. CAGR обсягів — індикатор реального попиту на дедоларизацію торгівлі.


```python
# JOIN: пов'язуємо нафтоюань з ціною нафти і світовим виробництвом
sql_q7 = f"""
SELECT
    p.year,
    ROUND(SUM(p.production_mbd), 2) AS world_production_mbd,
    ROUND(AVG(d.petroyuan_ine_futures_volume_klots), 0) AS petroyuan_volume,
    ROUND(AVG(d.oil_trade_non_usd_pct), 1) AS non_usd_trade_pct,
    ROUND(AVG(d.de_dollarization_pressure_index), 1) AS dedol_index
FROM {full_table_name('oil_production_trade')} p
JOIN {full_table_name('dedollarization')} d
    ON p.year = d.year AND p.month = d.month
GROUP BY p.year
ORDER BY p.year
"""

q7_petroyuan = run_query(sql_q7)
q7_petroyuan.tail(9)
```

```python
# Розраховуємо CAGR нафтоюаня з моменту запуску INE-ф'ючерсів (2018)
post_2018 = q7_petroyuan[q7_petroyuan["year"] >= 2018]
volumes = post_2018["petroyuan_volume"]

start_volume = volumes.iloc[0]
end_volume = volumes.iloc[-1]
last_year = int(post_2018["year"].iloc[-1])
n_years = len(post_2018) - 1

cagr = ((end_volume / start_volume) ** (1 / n_years) - 1) * 100

print(f"Обсяг нафтоюаня 2018: {start_volume:.0f} тис. лотів")
print(f"Обсяг нафтоюаня {last_year}: {end_volume:.0f} тис. лотів")
print(f"CAGR (2018 → {last_year}):   {cagr:.2f} % на рік")
```

```python
# Line chart: зростання нафтоюаня — лінійна та логарифмічна шкали
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Дані для обох subplots
years = q7_petroyuan["year"]
volume = q7_petroyuan["petroyuan_volume"]

# Лінійна шкала
ax1.fill_between(years, volume, alpha=0.25, color="#E53935")
ax1.plot(years, volume,
         color="#C62828", linewidth=2.5, marker="o", markersize=4)
ax1.axvline(2018, color="black", linestyle="--", alpha=0.5)
ax1.text(2018.2, volume.max() * 0.5,
         "Запуск INE\n(2018)", fontsize=9, color="black")
ax1.set_title("Нафтоюань: обсяги ф'ючерсів (лінійна шкала)")
ax1.set_xlabel("Рік")
ax1.set_ylabel("Тис. лотів")
ax1.xaxis.set_major_locator(MultipleLocator(5))
ax1.grid(False)

# Логарифмічна шкала
positive = q7_petroyuan[volume > 0]
ax2.semilogy(positive["year"], positive["petroyuan_volume"],
             color="#C62828", linewidth=2.5, marker="o", markersize=4)
ax2.axvline(2018, color="black", linestyle="--", alpha=0.5)
ax2.set_title(f"Нафтоюань: лог. шкала (CAGR з 2018 = {cagr:.0f} %/рік)")
ax2.set_xlabel("Рік")
ax2.set_ylabel("Тис. лотів (log)")
ax2.grid(False)

plt.tight_layout()
plt.show()
```

**Спостереження:** Нафтоюань демонструє вибуховий ріст: **CAGR = 59 % на рік** (з 56 тис. лотів у 2018 до 1431 тис. у 2025 — у 25 разів за 7 років). Дві шкали розповідають різні історії: на **лінійній** видно, що найбільший абсолютний приріст припадає на 2024-2025 роки (+200 тис. лотів щороку), а на **логарифмічній** — навпаки: найкрутіший нахил у перші роки після запуску INE (2018-2020), коли обсяги зросли в 7 разів за 2 роки. Після 2020 крива на логарифмічній шкалі **стає пологішою** — це класичний патерн зрілості нового ринку: ранні роки дають вибухові відсотки, далі темпи природно сповільнюються. ### Питання 9 — Чи статистично значуща різниця у ціні Brent між кризовими та стабільними роками?

**Бізнес-логіка:** Якщо кризи системно зрушують ціни — це треба враховувати в моделях прогнозування. Незалежний t-test перевірить різницю між двома вибірками.

- **H₀:** середня ціна Brent у кризові роки = у стабільні роки
- **H₁:** середні значення відрізняються (t-test)


```python
# t-test двох вибірок: кризові vs стабільні роки
crisis_prices = df_prices[
    df_prices["year"].isin(CRISIS_YEARS)
]["brent_crude_usd"]
stable_prices = df_prices[
    ~df_prices["year"].isin(CRISIS_YEARS)
]["brent_crude_usd"]

t_q9, p_q9 = stats.ttest_ind(crisis_prices, stable_prices)

print(f"Кризові роки:   середня ціна = ${crisis_prices.mean():.2f}/bbl, "
      f"n = {len(crisis_prices):,}")
print(f"Стабільні роки: середня ціна = ${stable_prices.mean():.2f}/bbl, "
      f"n = {len(stable_prices):,}")
print()
print(f"t-statistic: {t_q9:.3f}")
print(f"p-value:     {p_q9:.4e}")
print()

if p_q9 < 0.05:
    print("✅ Відхиляємо H₀: різниця цін статистично значуща")
else:
    print("❌ Не відхиляємо H₀: значущого зв'язку не виявлено")
```

```python
# Boxplot: розподіл ціни Brent у кризові vs стабільні роки
df_plot = df_prices.assign(
    period_type=np.where(
        df_prices["year"].isin(CRISIS_YEARS),
        "Кризові роки",
        "Стабільні роки",
    )
)

fig, ax = plt.subplots(figsize=(11, 5))
sns.boxplot(data=df_plot, x="period_type", y="brent_crude_usd",
            hue="period_type", palette=["#1565C0", "#E53935"],
            legend=False, ax=ax, width=0.5)
sns.stripplot(data=df_plot.sample(min(500, len(df_plot)), random_state=42),
              x="period_type", y="brent_crude_usd",
              alpha=0.2, color="black", size=2, ax=ax)

ax.set_title(
    f"Розподіл ціни Brent: кризові vs стабільні роки "
    f"(p = {p_q9:.2e})"
)
ax.set_xlabel("Період")
ax.set_ylabel("Ціна Brent, USD/bbl")
ax.grid(False)
plt.tight_layout()
plt.show()
```

**Спостереження:** Різниця у середніх цінах статистично **дуже значуща (p < 0.001)**: кризові роки в середньому коштують **\$59/bbl проти \$69/bbl у стабільні** — це майже на 15 % дешевше. Але цікаво **не лише середнє, а й форма розподілів**. Стабільні роки мають **широкий розкид** (Q1-Q3: \$51-\$85, повний діапазон \$20-\$120) — без жодних аутлаєрів: ринок проходить через різні режими (низькі ціни 2000-х, бум 2010-х, відновлення 2021-2022). Кризові роки натомість **дуже компактні** (Q1-Q3: \$44-\$64), але мають **виражений хвіст викидів зверху** до \$103/bbl — це короткі періоди різких відскоків під час криз (як-от середина 2008 перед обвалом [Lehman Brothers](https://uk.wikipedia.org/wiki/Lehman_Brothers), чи короткий пік 2020 перед локдаунами COVID-19).

### Питання 10 — Чи відрізняється дохід на барель між членами та нечленами ОПЕК?

**Бізнес-логіка:** Якщо членство в ОПЕК дає реальні переваги — члени мають отримувати більше доходу з одного барелю.

- **H₀:** дохід на барель ОПЕК = не ОПЕК
- **H₁:** значення відрізняються (t-test)


```python
# Ранжування: як дохід на барель відрізняється по країнам в межах кожної групи (ОПЕК/не ОПЕК)
sql_q10 = f"""
WITH country_revenue AS (
    SELECT
        country,
        is_opec_member,
        AVG(production_mbd) AS avg_production_mbd,
        AVG(oil_revenue_usd_bn) AS avg_revenue_usd_bn,
        -- Дохід на барель = місячний дохід (USD) / місячне виробництво (барель)
        -- 1e3: млрд → млн USD;   30: дні → місяць
        AVG(oil_revenue_usd_bn) * 1e3
            / (AVG(production_mbd) * 30) AS revenue_per_bbl
    FROM {full_table_name('oil_production_trade')}
    GROUP BY country, is_opec_member
)
SELECT
    country,
    is_opec_member,
    ROUND(avg_production_mbd, 2) AS avg_production_mbd,
    ROUND(revenue_per_bbl, 2) AS revenue_per_bbl,
    RANK() OVER (
        PARTITION BY is_opec_member
        ORDER BY revenue_per_bbl DESC
    ) AS rank_in_group
FROM country_revenue
ORDER BY is_opec_member DESC, rank_in_group
"""

q10_ranked = run_query(sql_q10)
q10_ranked
```

Розкид доходу на барель між групами дуже різний. **Серед членів ОПЕК діапазон вузький: \$40-61/bbl** (різниця 1.5 раза) — картель ефективно вирівнює умови для більшості учасників. Усі топ-7 країн за доходом на барель — члени ОПЕК. **Серед не ОПЕК ситуація драматично інша — діапазон \$2-56/bbl, різниця у 23 рази!** На вершині — **Норвегія (\$56)** і **Казахстан (\$54)**, які заробляють як середній член ОПЕК. **США — шокуючий аутсайдер з \$2.37/bbl**: при виробництві 8 Mb/d країна майже **не отримує експортного доходу**, бо більшість нафти переробляється і споживається всередині країни. Це парадокс «найбільшого виробника, який мало заробляє на нафті як на експортному товарі».

```python
# t-test двох вибірок: ОПЕК & не ОПЕК
df_prod_ext = df_prod.copy()
df_prod_ext["revenue_per_bbl"] = (
    df_prod_ext["oil_revenue_usd_bn"] * 1e3
    / (df_prod_ext["production_mbd"] * 30)
)

opec_rev = df_prod_ext[df_prod_ext["is_opec_member"]]["revenue_per_bbl"]
non_opec_rev = df_prod_ext[~df_prod_ext["is_opec_member"]]["revenue_per_bbl"]

t_q10, p_q10 = stats.ttest_ind(opec_rev, non_opec_rev)

print(f"ОПЕК:    середній дохід = ${opec_rev.mean():.2f}/bbl, "
      f"медіана = ${opec_rev.median():.2f}/bbl, n = {len(opec_rev):,}")
print(f"Не ОПЕК: середній дохід = ${non_opec_rev.mean():.2f}/bbl, "
      f"медіана = ${non_opec_rev.median():.2f}/bbl, n = {len(non_opec_rev):,}")
print()
print(f"t-statistic: {t_q10:.3f}")
print(f"p-value:     {p_q10:.4e}")
print()

if p_q10 < 0.05:
    print("✅ Відхиляємо H₀: різниця доходу на барель статистично значуща")
else:
    print("❌ Не відхиляємо H₀: значущого зв'язку не виявлено")
```

```python
# Boxplot: розподіл доходу на барель між країнами ОПЕК vs не ОПЕК
fig, ax = plt.subplots(figsize=(11, 5))

box_data = [opec_rev, non_opec_rev]
labels = ["Члени ОПЕК", "Нечлени ОПЕК"]
colors = ["#E53935", "#1565C0"]

bp = ax.boxplot(box_data, labels=labels, patch_artist=True, widths=0.5)
for patch, color in zip(bp["boxes"], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.6)

# Додаємо середні як трикутники
ax.scatter([1, 2], [opec_rev.mean(), non_opec_rev.mean()],
           marker="^", color="black", s=80, zorder=5, label="Середнє")

ax.set_title(f"Дохід на барель: ОПЕК vs не ОПЕК (p = {p_q10:.2e})")
ax.set_ylabel("Дохід на барель, USD")
ax.legend(loc="upper right")
ax.grid(False)
plt.tight_layout()
plt.show()
```

**Спостереження:** Різниця статистично високо значуща: **члени ОПЕК отримують у середньому \$55/bbl, нечлени \$39/bbl** (p < 0.001) — це підтверджує, що **членство в картелі дає реальну економічну перевагу** приблизно на \$16/bbl. Проте графік показує важливий нюанс: **бокси частково перекриваються** — верхня третина не ОПЕК (≈\$55-105) збігається з нижньою третиною ОПЕК. Це означає, що **бути в ОПЕК не гарантує найвищого доходу**. Не ОПЕК має **значно ширший розкид (\$1-105 vs \$13-115)**, що підтверджує висновок вище: ОПЕК — стандартизована монетизація, не ОПЕК = індивідуальні стратегії від США (≈\$2/bbl, переробка вдома) до Норвегії (≈\$56/bbl, преміум-експорт). Застосування SQL (Window Function) допомогла побачити це ранжування всередині кожної групи окремо.

```python
crisis_events = {
    "2008-09-15": ("Lehman Brothers", "top"),
    "2014-11-01": ("ОПЕК: цінова війна", "top"),
    "2020-04-01": ("Локдауни COVID-19", "bottom"),
    "2022-02-24": ("Вторгнення в Україну", "top"),
}

fig, ax = plt.subplots(figsize=(14, 5))

ax.plot(df_prices["date"], df_prices["brent_crude_usd"],
        linewidth=1.3, color="#1565C0", alpha=0.9, label="Brent (USD/bbl)")
ax.fill_between(df_prices["date"], df_prices["brent_crude_usd"],
                alpha=0.08, color="#1565C0")

for date_str, (label, position) in crisis_events.items():
    event_date = pd.Timestamp(date_str)
    idx = (df_prices["date"] - event_date).abs().idxmin()
    price_at_event = df_prices.loc[idx, "brent_crude_usd"]
    ax.axvline(event_date, color="red", linestyle="--",
               alpha=0.4, linewidth=1.2)
    offset = 20 if position == "top" else -22
    ax.annotate(
        label, xy=(event_date, price_at_event),
        xytext=(event_date, price_at_event + offset),
        fontsize=9, ha="center", color="darkred",
        arrowprops=dict(arrowstyle="->", color="red", lw=1),
    )

ax.set_title("25 років цін Brent: ключові події, що змінили ринок нафти")
ax.set_xlabel("Дата")
ax.set_ylabel("Ціна Brent, USD/bbl")
ax.xaxis.set_major_locator(YearLocator(2))
ax.xaxis.set_major_formatter(DateFormatter("%Y"))
ax.grid(False)
plt.tight_layout()
plt.show()
```

**Що бачимо за 25 років:** Графік чітко демонструє, що ціна нафти — це **не випадковий шум, а результат взаємодії циклічних факторів і геополітичних подій**. Чотири аннотовані події утворюють чотири переломні моменти: Lehman Brothers 2008 закрив перший суперцикл, цінова війна ОПЕК 2014 запустила еру конкуренції зі сланцем, COVID-19 2020 дав найбільший шок попиту в історії, а вторгнення в Україну 2022 започаткувало нову еру дедоларизації. Між цими шоками видно класичний патерн: **повільне зростання → раптовий обвал → довге відновлення**, причому з часом цикл прискорюється: у 2008-2011 відновлення зайняло 3 роки, у 2020 — менше року.

### Підсумок

За 25 років нафтодоларова система зазнала **двох структурних шоків**: сланцева революція США (2010-х) підірвала монополію ОПЕК на пропозицію, а санкційна політика після 2014-2022 років прискорила пошук альтернатив долару у міжнародних розрахунках. Однак обидва процеси **поступові**, а не революційні: ОПЕК зберігає економічну перевагу (\$55 проти \$39 на барель), а долар залишається валютою приблизно 57 % глобальних резервів. **Нафтодолар не помер — але він уже не такий всемогутній, як був у 2000-х**. Це не точка слабкості системи, а сигнал її **поступової трансформації** від моноцентричної до поліцентричної архітектури глобальних фінансів.
